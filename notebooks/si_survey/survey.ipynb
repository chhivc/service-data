```python
import pandas as pd
```

```python
rawdata = pd.read_csv("si_survey_all_responses_rawdata.csv")
```

```python
col_labels = rawdata.columns
for i in range(len(col_labels)):
    label = col_labels[i].replace("\n", "")
    print(f"{i}   {label}")
```

```python
#rawdata.head(5)
```

```python
# Rename column label
surveyData = rawdata.rename(
    columns={
        col_labels[0]: col_labels[0].split("/")[0].strip(),
        col_labels[1]: col_labels[1].split("/")[0].strip(),
        col_labels[2]: col_labels[2].replace("\n", ""),
        col_labels[3]: col_labels[3].replace("\n", ""),
        col_labels[4]: col_labels[4].replace("\n", ""),
        col_labels[5]: col_labels[5].replace("\n", ""),
        col_labels[6]: col_labels[6].replace("\n", ""),
        col_labels[7]: col_labels[7][:4],
        col_labels[8]: col_labels[8][0],
        col_labels[9]: col_labels[9][0],
        col_labels[10]: col_labels[10][0],
        col_labels[11]: col_labels[11][:2],
        col_labels[12]: col_labels[12][:2],
        col_labels[13]: col_labels[13][:2],
        col_labels[14]: col_labels[14][:2],
        col_labels[15]: col_labels[14][:2] + "_other",
        col_labels[16]: col_labels[16][:2],
        col_labels[17]: col_labels[16][:2] + "_other",
        col_labels[18]: col_labels[18][:2],
    }
)
```
