```python
# Service standard data
ss_data = pd.read_csv("/workspaces/service-data/outputs/snapshots/2025-03-01/utils/ss_all.csv", sep=';')
```

```python
ss_data.count()
```

```python
ss23 = ss_data.query("fiscal_yr == '2023-2024'").copy()
ss22 = ss_data.query("fiscal_yr == '2022-2023'").copy()
```

```python
ss23.shape
```

```python
ss23_wtarget = ss23[~ss23["target"].isna()]
ss23_wtarget.shape
```

```python
ss23_notarget = ss23[ss23["target"].isna()]
ss23_notarget.shape
```

```python
target_cols = ["fiscal_yr", "service_id", "service_name_en", "service_standard_id", "service_standard_en", "type", "target_type", "target", "volume_meeting_target", "total_volume", "performance", "target_met", "department_en"]
ss23_notarget[target_cols].head(10)
```

```python
# Service standards matched to extern/enterprise services
id_si23_ext = set(si23_ext["service_id"].values)
ss23_notarget_ext = ss23_notarget[ss23_notarget["service_id"].isin(id_si23_ext)]
ss23_notarget_ext.shape
```

```python
ss23_notarget_vol = ss23_notarget[~ss23_notarget["total_volume"].isna()]
ss23_notarget_vol.shape
```

```python
ss23_notarget_vol[target_cols][:10]
```

```python
ss23_notarget_vol_vmt = ss23_notarget_vol[~ss23_notarget_vol["volume_meeting_target"].isna()]
ss23_notarget_vol_vmt.shape
```

```python
ss23_notarget_vol_bydept = ss23_notarget_vol.groupby("department_en").size().reset_index(name="number_ss")
ss23_notarget_vol_bydept = ss23_notarget_vol_bydept.sort_values(by=["number_ss"], ascending=False)
ss23_notarget_vol_bydept[:10]
```

```python
# Standards with no target by department
ss23_notarget_bydept = ss23_notarget.groupby("department_en").size().reset_index(name="number_ss")
ss23_notarget_bydept = ss23_notarget_bydept.sort_values(by=["number_ss"], ascending=False)
ss23_notarget_bydept[:10]
```

```python
ss23_notarget_vol.to_csv("service_standards_no_target_with_volume.csv", sep=";")
```

```python
# Standards with both volume empty
ss_both_volume_empty = ss23_notarget[ss23_notarget["volume_meeting_target"].isna() & ss23_notarget["total_volume"].isna()]
ss_both_volume_empty.shape
```

```python
#ss_both_volume_empty[target_cols][:10]
```

```python
# Standards with only total volume empty but volume_meeting_target is filled
ss_totalVolume_empty = ss23_notarget[~ss23_notarget["volume_meeting_target"].isna() & ss23_notarget["total_volume"].isna()]
ss_totalVolume_empty.shape
```

```python
# Standards with total volume empty fillled but volume_meeting_target empty
ss_volMeetTarget_empty = ss23_notarget[ss23_notarget["volume_meeting_target"].isna() & ~ss23_notarget["total_volume"].isna()]
ss_volMeetTarget_empty.shape
```

```python
# Standard with total volume == zero
ss_totalVolume_zero = ss23_notarget[ss23_notarget["total_volume"] == 0]
ss_totalVolume_zero.shape
```

```python
# standards with total volume > zero and volume meeting target is zero or greater
ss_have_both_volume = ss23_notarget[~ss23_notarget["volume_meeting_target"].isna() & ~ss23_notarget["total_volume"].isna() & ss23_notarget["total_volume"] > 0]
ss_have_both_volume.shape
```

```python
ss_both_volume_empty.to_csv("Empty_volume_fields.csv", sep=(";"))
ss_volMeetTarget_empty.to_csv("Only_total_volume_filled.csv", sep=";")
ss_totalVolume_zero.to_csv("Total_volume_zero.csv", sep=";")
ss_have_both_volume.to_csv("Have both_volumes.csv", sep=";")
```
